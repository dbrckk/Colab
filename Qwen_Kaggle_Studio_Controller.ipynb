# Qwen Kaggle Studio — Contrôleur léger

Ce notebook **ne fait pas la génération lourde**. Il lance uniquement l'interface Gradio, envoie les jobs à Kaggle GPU, récupère les résultats et les stocke dans Google Drive.

Après lancement, ouvre l'URL Gradio affichée. Les identifiants Kaggle peuvent être saisis dans l'onglet **Paramètres Kaggle**.


```python
#@title ▶ Lancer Qwen Kaggle Studio
from google.colab import drive
drive.mount('/content/drive')

import os, subprocess, sys, pathlib, shutil

REPO_URL = 'https://github.com/dbrckk/Colab.git'
REPO_DIR = pathlib.Path('/content/Colab')
PERSIST = pathlib.Path('/content/drive/MyDrive/QwenKaggleStudio')
PERSIST.mkdir(parents=True, exist_ok=True)

os.environ['QWEN_KAGGLE_STORAGE'] = str(PERSIST / 'media')
os.environ['QWEN_KAGGLE_DB'] = str(PERSIST / 'jobs.sqlite3')
os.environ['QWEN_KAGGLE_ENV_FILE'] = str(PERSIST / '.env.local')
os.environ['QWEN_KAGGLE_SHARE'] = 'true'

if REPO_DIR.exists() and (REPO_DIR / '.git').exists():
    subprocess.run(['git','-C',str(REPO_DIR),'fetch','origin','main'], check=True)
    subprocess.run(['git','-C',str(REPO_DIR),'reset','--hard','origin/main'], check=True)
else:
    shutil.rmtree(REPO_DIR, ignore_errors=True)
    subprocess.run(['git','clone','--depth','1',REPO_URL,str(REPO_DIR)], check=True)

os.chdir(REPO_DIR)
print('Stockage persistant :', PERSIST)
subprocess.run([sys.executable, 'launch_kaggle_ui.py'], check=True)
```
